# Static drawing (py5bot kernel)

Each cell of a **py5bot** notebook is a complete static sketch: write drawing commands top to bottom (imported mode, no `setup()`/`draw()`), and the cell's output is the image. Good for quick experiments and generative stills; needs no visible window at all.

```python
size(400, 400)
background(20, 24, 33)
no_fill()
stroke_weight(2)
random_seed(5)
for i in range(24):
    stroke(80 + 7 * i, 160, 255 - 8 * i)
    circle(200 + random(-30, 30), 200 + random(-30, 30), 20 + 14 * i)
```

```python
size(400, 200)
background(255)
no_stroke()
for x in range(0, 400, 20):
    fill(x * 255 / 400, 120, 200)
    rect(x, 100 - x / 4, 18, x / 2)
```
