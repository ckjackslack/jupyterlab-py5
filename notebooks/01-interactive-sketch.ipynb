# Interactive sketch (py5 kernel)

The original example: squares follow the mouse, clicking changes the fill colour.

The **py5** kernel runs in *imported mode*, so `size()`, `square()`, `mouse_x` … need no `py5.` prefix and `run_sketch()` returns immediately, leaving the notebook usable while the sketch runs.

Inside the Docker container the sketch window opens on a virtual (Xvfb) display that you cannot see. Use `py5_tools.screenshot()` (below) to pull frames into the notebook, or use the X11 overlay (`compose.x11.yaml`, see README) to get the real, clickable window on a Linux desktop. When no mouse is over the window the squares orbit the centre instead, so there is always something to see.

```python
import math


def setup():
    size(400, 400)
    rect_mode(CENTER)
    background(255)


def draw():
    if focused and (mouse_x or mouse_y):
        x, y = mouse_x, mouse_y
    else:  # nobody is moving the mouse (e.g. headless): orbit the centre
        t = frame_count / 15
        x, y = 200 + 140 * math.cos(t), 200 + 140 * math.sin(t * 1.5)
    square(x, y, 10)


def mouse_clicked():
    fill(random_int(255), random_int(255), random_int(255))


run_sketch()
```

Grab what the sketch has drawn so far. Run this cell again to see it progress; call `mouse_clicked()` yourself to change the colour without a mouse.

```python
import time

import py5_tools

time.sleep(1.5)  # let it draw for a moment
mouse_clicked()
time.sleep(1.5)
py5_tools.screenshot()
```
