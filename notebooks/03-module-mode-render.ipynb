# Module mode in a plain Python kernel

You do not need the py5 kernels at all: `import py5` in the regular Python kernel and use the render helpers, which run a hidden sketch and return PIL images. Handy when mixing py5 with numpy, pandas or matplotlib.

```python
import py5


@py5.render(300, 300)
def rings(s: py5.Sketch) -> None:
    s.background(250)
    s.no_fill()
    for i in range(12):
        s.stroke(30, 60 + 15 * i, 200)
        s.circle(150, 150, 20 * i + 10)


rings()
```

`render_sequence` returns a list of frames; here they are tiled into a contact sheet.

```python
from PIL import Image


def setup(s: py5.Sketch) -> None:
    s.rect_mode(py5.CENTER)


def draw(s: py5.Sketch) -> None:
    s.background(255)
    s.translate(60, 60)
    s.rotate(s.frame_count * py5.PI / 16)
    s.fill(255, 120, 40)
    s.square(0, 0, 60)


frames = py5.render_frame_sequence(draw, 120, 120, setup=setup, limit=8)
sheet = Image.new("RGB", (120 * len(frames), 120))
for i, frame in enumerate(frames):
    sheet.paste(frame, (120 * i, 0))
sheet
```
